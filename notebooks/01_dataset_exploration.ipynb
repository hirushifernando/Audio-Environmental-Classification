```python
from datasets import load_dataset

dataset = load_dataset(
    "renumics/esc50",
    split="train"
)

dataset
```

```python
print(dataset.features)
```

```python
label_names = dataset.features["label"].names

for i, label in enumerate(label_names):
    print(i, label)
```

```python
print("Number of classes:", len(label_names))
```

```python
from collections import Counter

label_counts = Counter(dataset["label"])

for label_id, count in sorted(label_counts.items()):
    print(label_names[label_id], ":", count)
```

```python
selected_classes = [
    "dog",
    "cat",
    "rain",
    "clapping",
    "footsteps",
    "car_horn",
    "siren",
    "thunderstorm",
    "keyboard_typing",
    "church_bells"
]

selected_ids = [
    label_names.index(name)
    for name in selected_classes
]

selected_ids
```

```python
selected_indices = [
    i for i, label_id in enumerate(dataset["label"])
    if label_id in selected_ids
]

print("Selected recordings:", len(selected_indices))
```

```python
small_dataset = dataset.select(selected_indices)

print(small_dataset)
```

```python
small_dataset = small_dataset.shuffle(seed=42)
```

```python
small_dataset = small_dataset.select(
    range(min(250, len(small_dataset)))
)
```

```python
print(small_dataset)
```

```python
print("Number of recordings:", len(small_dataset))
print("Columns:", small_dataset.column_names)
```

```python
from collections import Counter

label_counts = Counter(small_dataset["label"])

for label_id, count in label_counts.items():
    print(label_names[label_id], ":", count)
```

```python
src_file = small_dataset["src_file"][0]
label_id = small_dataset["label"][0]

print("File:", src_file)
print("Label ID:", label_id)
print("Label:", label_names[label_id])
```
